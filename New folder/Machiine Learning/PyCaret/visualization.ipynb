```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
```

```python
df = pd.read_csv('heart_disease_uci.csv')
df.head()
```

# 1. Scatter plot

```python
fig = px.scatter(df, x='age', y='thalch', color='sex', symbol='sex', title='thalch ratio in age').show()
```

```python
df['e'] = df["age"]/100
fig = px.scatter(df, x='age', y='thalch', color='sex', error_x='e', error_y='e').show()
```

```python
fig = px.scatter(df, x='age', y='thalch', marginal_x='histogram', marginal_y='rug').show()
```

```python
fig =  px.scatter(df, x='age', y='thalch', trendline='ols').show()
```

```python
df = px.scatter(df, x='age', y='thalch', color='sex', title='thalch ratio in age', size='num', hover_data=['num']).show()
```

# 2. Lineplot

```python
fig_mean = df.groupby('age')['thalch'].mean().reset_index()
fig = px.line(fig_mean, x='age', y='thalch').show()
```
